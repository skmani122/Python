CONCEPT: The Sliding Window Technique
-------------------------------------
Sliding window is used when you need to examine CONTIGUOUS subarrays/substrings
and calculate something (sum, max, average, count, etc.).

Instead of recalculating from scratch for each position, you:
    1. Build the initial window
    2. SLIDE it forward by adding one element and removing one element

This turns O(n × k) brute force into O(n) — a huge win for large inputs!


VISUAL: Fixed Window of Size 3
------------------------------
Array: [2, 1, 5, 1, 3, 2]
Find maximum sum of any 3 consecutive elements.

    [2, 1, 5, 1, 3, 2]
     └──┬──┘              Window 1: sum = 8
        └──┬──┘           Window 2: sum = 7  (subtract 2, add 1)
           └──┬──┘        Window 3: sum = 9  (subtract 1, add 3) ← MAX
              └──┬──┘     Window 4: sum = 6  (subtract 5, add 2)

Key insight: Each slide only changes 2 elements!
    new_sum = old_sum - element_leaving + element_entering


WHY SLIDING WINDOW BEATS BRUTE FORCE
------------------------------------
Brute force (recalculate each window):
    for i in range(n - k + 1):
        window_sum = sum(nums[i:i+k])  # O(k) each time!
    # Total: O(n × k)

Sliding window (update incrementally):
    # Calculate first window: O(k)
    # Slide n-k times, O(1) each: O(n)
    # Total: O(n + k) = O(n)

For k=1000 and n=1,000,000:
    Brute force: ~1 billion operations
    Sliding window: ~1 million operations (1000x faster!)


PYTHON TIP: float('-inf') and float('inf')
------------------------------------------
When finding maximum or minimum values, initialize with extreme values:

    # Finding maximum - start impossibly low
    max_val = float('-inf')  # Negative infinity

    # Finding minimum - start impossibly high
    min_val = float('inf')   # Positive infinity

Why not use 0?
    nums = [-5, -3, -1]
    max_val = 0  # Bug! Would return 0, but max is -1
    max_val = float('-inf')  # Correct! Any number beats -inf

Why not use nums[0]?
    You can! But float('-inf') works even for empty arrays
    and simplifies code when your loop structure varies.


PYTHON TIP: The sum() Function with Slices
------------------------------------------
Python's sum() is clean for the initial window:

    # Initial window sum
    window_sum = sum(nums[:k])  # Sum of first k elements

    # Or sum any slice
    sum(nums[2:5])  # Sum of indices 2, 3, 4

But DON'T use sum() inside your sliding loop — that defeats the purpose!

    # ❌ BAD: Recalculating entire sum each time
    for i in range(n - k + 1):
        current = sum(nums[i:i+k])  # O(k) each iteration!

    # ✅ GOOD: Update incrementally
    for i in range(k, n):
        window_sum += nums[i] - nums[i - k]  # O(1) each iteration!

TODAY'S PROBLEM: Maximum Sum Subarray of Size K
================================================================================

Given an array of integers and a number k, find the maximum sum of any
contiguous subarray of size k.

Requirements:
    - Use the sliding window technique
    - Return the maximum sum
    - If the array has fewer than k elements, return 0

Example:
    max_sum_subarray([2, 1, 5, 1, 3, 2], k=3) → 9
    # Subarray [5, 1, 3] has the maximum sum

    max_sum_subarray([2, 3, 4, 1, 5], k=2) → 7
    # Subarray [3, 4] has the maximum sum

Think about:
    - How do you build the initial window?
    - What's the formula for sliding the window?
    - When do you update max_sum?

```python
def max_sum_subarray(nums: list[int], k: int) -> int:
    """
    Find the maximum sum of any contiguous subarray of size k.

    Args:
        nums: List of integers
        k: Size of the window

    Returns:
        Maximum sum of any k consecutive elements, or 0 if len(nums) < k
    """
    if len(nums) < k:
        return 0
        
    max_sum = float('-inf')
    window_sum = sum(nums[:k])
    max_sum = window_sum

    for i in range(1, len(nums) - k + 1):
        if (k <= 3): # For small k, it's more efficient to calculate the sum directly
            print("Smaller k, calculating sum directly")
            window_sum = sum(nums[i:i+k])
        else:
            print("Larger k, using sliding window", window_sum)
            window_sum += nums[i + k - 1] - nums[i - 1]
        max_sum = max(max_sum, window_sum)

    return max_sum


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def test(run_num, input_list, k, expected):
    print(f"\nTest {run_num}: max_sum_subarray({input_list}, k={k})")
    result = max_sum_subarray(input_list, k=k)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: max_sum_subarray()")
    print("=" * 60)

    # Test 1: Standard case
    test(run_num=1, input_list=[2, 1, 5, 1, 3, 2], k=3, expected=9)

    # Test 2: Window size 2
    test(run_num=2, input_list=[2, 3, 4, 1, 5], k=2, expected=7)

    # Test 3: Max at end
    test(run_num=3, input_list=[1, 2, 3, 4, 5], k=3, expected=12)

    # Test 4: Max at beginning
    test(run_num=4, input_list=[5, 4, 3, 2, 1], k=2, expected=9)

    # Test 5: Window equals array length
    test(run_num=5, input_list=[1, 2, 3], k=3, expected=6)

    # Test 6: With negative numbers
    test(run_num=6, input_list=[-1, 2, 3, -2, 5], k=4, expected=8)

    # Test 7: All negative numbers
    test(run_num=7, input_list=[-3, -1, -4, -2], k=2, expected=-4)

    # Test 8: Array smaller than k
    test(run_num=8, input_list=[1, 2], k=5, expected=0)

    # Test 9: Empty array
    test(run_num=9, input_list=[], k=3, expected=0)

    # Test 10: Window size 1
    test(run_num=10, input_list=[3, 1, 4, 1, 5], k=1, expected=5)

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Visit each element once")
    print("  Space: O(1) - Only tracking sum and max (no extra arrays)")
    print("=" * 60)

run_tests()
```
