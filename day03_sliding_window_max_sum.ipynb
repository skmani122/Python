Day 3: Sliding Window Technique - Maximum Sum Subarray
=======================================================

CONCEPT: The sliding window technique is perfect for problems involving 
contiguous sequences (subarrays, substrings). Instead of recalculating 
from scratch for each window position, you "slide" the window by removing 
one element and adding another.

This transforms O(n*k) brute force into O(n) efficiency!

PYTHON TIP: The Walrus Operator (:=) and sum() with slices
----------------------------------------------------------
Python 3.8+ introduced the walrus operator for assignment expressions:

    # Without walrus - need separate lines
    line = input()
    while line != "quit":
        process(line)
        line = input()
    
    # With walrus - assign and check in one expression
    while (line := input()) != "quit":
        process(line)

    # Useful in list comprehensions
    results = [y for x in data if (y := expensive_function(x)) > threshold]

Also remember sum() works great with slices:
    
    arr = [1, 2, 3, 4, 5]
    sum(arr[0:3])  # Returns 6 (sum of first 3 elements)
    sum(arr[-2:])  # Returns 9 (sum of last 2 elements)

SLIDING WINDOW WALKTHROUGH:
    Array: [2, 1, 5, 1, 3, 2], k = 3
    
    Step 1: Calculate initial window sum
    [2, 1, 5] 1, 3, 2  → window_sum = 8
    
    Step 2: Slide window - subtract left element, add right element
     2 [1, 5, 1] 3, 2  → window_sum = 8 - 2 + 1 = 7
    
    Step 3: Continue sliding
     2, 1 [5, 1, 3] 2  → window_sum = 7 - 1 + 3 = 9  ← maximum!
    
    Step 4: Continue sliding
     2, 1, 5 [1, 3, 2] → window_sum = 9 - 5 + 2 = 6
    
    Answer: 9

TIME COMPLEXITY: O(n) - single pass through array
SPACE COMPLEXITY: O(1) - only storing a few variables

```python
def max_sum_subarray(arr: list, k: int) -> int:
    """
    Find the maximum sum of any contiguous subarray of size k.
    
    Use the sliding window technique:
    1. Calculate the sum of the first window (first k elements)
    2. Slide the window: subtract the element leaving, add the element entering
    3. Track the maximum sum seen
    
    Args:
        arr: List of integers
        k: Size of the subarray window
    
    Returns:
        Maximum sum of any subarray of size k
        Returns 0 if array is empty or k is larger than array length
    
    Hints:
    - Handle edge cases first (empty array, k > len(arr))
    - sum(arr[:k]) gives you the first window sum
    - As you slide: new_sum = old_sum - arr[i-k] + arr[i]
    """
    if not arr or k > len(arr):
        return 0

    max_sum = current_sum = sum(arr[:k])

    for i in range(k, len(arr)):
        current_sum += arr[i] - arr[i - k]

        # Find the maximum sum
        #max_sum = max(max_sum, current_sum)
        # Alternatively, you could use walrus operator (Python 3.8+):
        if (current := current_sum) > max_sum:
            max_sum = current

    return max_sum

# ============ TESTS ============
if __name__ == "__main__":
    print("Testing max_sum_subarray...")
    print("=" * 50)

    # Test 1: Basic case
    result = max_sum_subarray([2, 1, 5, 1, 3, 2], 3)
    print(f"Test 1 - [2, 1, 5, 1, 3, 2], k=3")
    print(f"  Expected: 9, Got: {result}: {'✓ PASS' if result == 9 else '✗ FAIL'}")

    # Test 2: Maximum at the beginning
    result = max_sum_subarray([5, 4, 3, 2, 1], 2)
    print(f"Test 2 - [5, 4, 3, 2, 1], k=2")
    print(f"  Expected: 9, Got: {result}: {'✓ PASS' if result == 9 else '✗ FAIL'}")

    # Test 3: Maximum at the end
    result = max_sum_subarray([1, 2, 3, 4, 5], 2)
    print(f"Test 3 - [1, 2, 3, 4, 5], k=2")
    print(f"  Expected: 9, Got: {result}: {'✓ PASS' if result == 9 else '✗ FAIL'}")

    # Test 4: Window size equals array length
    result = max_sum_subarray([1, 2, 3], 3)
    print(f"Test 4 - [1, 2, 3], k=3")
    print(f"  Expected: 6, Got: {result}: {'✓ PASS' if result == 6 else '✗ FAIL'}")

    # Test 5: Window size of 1
    result = max_sum_subarray([3, 1, 4, 1, 5, 9, 2, 6], 1)
    print(f"Test 5 - [3, 1, 4, 1, 5, 9, 2, 6], k=1")
    print(f"  Expected: 9, Got: {result}: {'✓ PASS' if result == 9 else '✗ FAIL'}")

    # Test 6: Negative numbers
    result = max_sum_subarray([-1, -2, -3, -4], 2)
    print(f"Test 6 - [-1, -2, -3, -4], k=2")
    print(f"  Expected: -3, Got: {result}: {'✓ PASS' if result == -3 else '✗ FAIL'}")

    # Test 7: Mixed positive and negative
    result = max_sum_subarray([2, -1, 3, -2, 4, -3], 3)
    print(f"Test 7 - [2, -1, 3, -2, 4, -3], k=3")
    print(f"  Expected: 5, Got: {result}: {'✓ PASS' if result == 5 else '✗ FAIL'}")

    # Test 8: Empty array (edge case)
    result = max_sum_subarray([], 3)
    print(f"Test 8 - [], k=3")
    print(f"  Expected: 0, Got: {result}: {'✓ PASS' if result == 0 else '✗ FAIL'}")

    # Test 9: k larger than array (edge case)
    result = max_sum_subarray([1, 2], 5)
    print(f"Test 9 - [1, 2], k=5")
    print(f"  Expected: 0, Got: {result}: {'✓ PASS' if result == 0 else '✗ FAIL'}")

    print("=" * 50)
    print("All tests complete!")
    print("\n")
```
