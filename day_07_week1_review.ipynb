PYTHON TIP: max() with Multiple Arguments
-----------------------------------------
Python's max() is flexible:

    # Compare two values
    max(5, 3)  # → 5

    # Compare multiple values
    max(3, 1, 4, 1, 5)  # → 5

    # Find max in a list
    max([3, 1, 4, 1, 5])  # → 5

    # Update a running maximum
    best = 0
    best = max(best, current)  # Keep the larger one

    # With key function (advanced)
    words = ["apple", "pie", "strawberry"]
    max(words, key=len)  # → "strawberry" (longest)


PYTHON TIP: Calculate Area/Distance Cleanly
-------------------------------------------
For geometry-style problems, break down the calculation:

    # Width between two indices
    width = right - left

    # Height (often the minimum of two values)
    height = min(heights[left], heights[right])

    # Area
    area = width * height

    # One-liner version
    area = (right - left) * min(heights[left], heights[right])

Both are O(1). Use whichever is clearer to you.

================================================================================
TODAY'S PROBLEM: Container With Most Water
================================================================================

You are given an array of positive integers where each element represents
the height of a vertical line at that position. Find two lines that together
with the x-axis form a container that holds the most water.

The container's water capacity is: width × min(height1, height2)
- width = distance between the two lines (difference in indices)
- height = the shorter of the two lines (water overflows the shorter one)

Requirements:
- Return the maximum amount of water the container can hold
- You must use the two-pointer approach (not brute force)

Example:
    max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]) → 49
    # Lines at index 1 (height=8) and index 8 (height=7)
    # Width = 8 - 1 = 7, Height = min(8, 7) = 7
    # Area = 7 × 7 = 49

Visual:
        8           8
        |   6       |
        |   |   5   |   7
        |   |   | 4 |   |
        |   | 2 | | |   | 3
    1   |   | | | | |   | |
    |   |   | | | | |   | |
    ─────────────────────────
    0   1   2 3 4 5 6   7 8

Think about:
    - Why start with pointers at opposite ends?
    - Which pointer should you move? The one at the shorter or taller line?
    - Why does moving the shorter pointer give you the best chance of improvement?


```python
def max_area(heights: list[int]) -> int:
    """
    Find the maximum water container area formed by two lines.

    Args:
        heights: List of positive integers representing line heights

    Returns:
        Maximum area of water that can be contained
    """
    left, right = 0, len(heights) - 1
    max_area = 0

    while left < right:
        # Calculate area with current pointers
        width = right - left
        height = min(heights[left], heights[right])
        current_area = width * height
        max_area = max(max_area, current_area)

        # Move the pointer of the shorter line inward
        if heights[left] < heights[right]:
            left += 1
        else:
            right -= 1

    return max_area


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def test(run_num: int, input_list: list[int], expected: int):
    result = max_area(input_list)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest {run_num}: max_area({input_list})")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: max_area()")
    print("=" * 60)

    # Test 1: Standard case (LeetCode example)
    test(run_num=1, input_list=[1, 8, 6, 2, 5, 4, 8, 3, 7], expected=49)

    # Test 2: Two elements
    test(run_num=2, input_list=[1, 1], expected=1)

    # Test 3: Increasing heights
    test(run_num=3, input_list=[1, 2, 3, 4, 5], expected=6)

    # Test 4: Decreasing heights
    test(run_num=4, input_list=[5, 4, 3, 2, 1], expected=6)

    # Test 5: All same height
    test(run_num=5, input_list=[3, 3, 3, 3], expected=9)

    # Test 6: Tall ends, short middle
    test(run_num=6, input_list=[10, 1, 1, 1, 10], expected=40)

    # Test 7: Short ends, tall middle
    test(run_num=7, input_list=[1, 10, 10, 1], expected=10)

    # Test 8: Larger example
    test(run_num=8, input_list=[2, 3, 10, 5, 7, 8, 9], expected=36)

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Each pointer moves at most n times total")
    print("  Space: O(1) - Only storing pointers and max_area")
    print("=" * 60)
    print("\n  Brute Force would be O(n²) - checking all pairs.")
    print("  Two pointers reduces this to O(n)!")

if __name__ == "__main__":
    run_tests()
```
