CONCEPT: Sets — The "Unique Collection" Data Structure
------------------------------------------------------
A set is an unordered collection of UNIQUE elements. Duplicates are impossible.

    nums = [1, 2, 2, 3, 3, 3]
    unique = set(nums)  # {1, 2, 3}

Sets are backed by hash tables, so they have O(1) membership testing:

    3 in unique  # O(1) - just like dict keys!


SET OPERATIONS — Think Venn Diagrams!
-------------------------------------
Sets support mathematical set operations:

    a = {1, 2, 3, 4}
    b = {3, 4, 5, 6}

    # UNION — all elements from both sets
    a | b            # {1, 2, 3, 4, 5, 6}
    a.union(b)       # same thing

    # INTERSECTION — elements in BOTH sets
    a & b            # {3, 4}
    a.intersection(b)

    # DIFFERENCE — elements in a but NOT in b
    a - b            # {1, 2}
    a.difference(b)

    # SYMMETRIC DIFFERENCE — elements in either but NOT both
    a ^ b            # {1, 2, 5, 6}
    a.symmetric_difference(b)

These operations are very efficient and often turn O(n²) into O(n)!


SET OPERATIONS COMPLEXITY
-------------------------
    s = {1, 2, 3}

    # O(1) operations
    s.add(4)         # Add element
    s.remove(3)      # Remove (raises KeyError if missing)
    s.discard(3)     # Remove (no error if missing)
    3 in s           # Membership check
    len(s)           # Size

    # O(n) operations
    s.pop()          # Remove arbitrary element
    set(my_list)     # Create set from list

    # O(len(a) + len(b)) for set operations
    a | b            # Union
    a & b            # Intersection
    a - b            # Difference


PYTHON TIP: Creating Sets
-------------------------
    # From a list (removes duplicates)
    set([1, 2, 2, 3])  # {1, 2, 3}

    # Set literal (curly braces)
    {1, 2, 3}          # {1, 2, 3}

    # Empty set — CAREFUL! {} creates an empty DICT, not set!
    empty_dict = {}    # This is a dict!
    empty_set = set()  # This is a set!

    # Set comprehension
    {x**2 for x in range(5)}  # {0, 1, 4, 9, 16}


PYTHON TIP: Converting Set Back to List
---------------------------------------
Sets are unordered, so convert back to list if you need order or indexing:

    my_set = {3, 1, 2}
    my_list = list(my_set)      # [1, 2, 3] or any order!
    sorted_list = sorted(my_set) # [1, 2, 3] guaranteed sorted


PYTHON TIP: Checking Subsets and Supersets
------------------------------------------
    a = {1, 2}
    b = {1, 2, 3, 4}

    a.issubset(b)    # True — all of a's elements are in b
    a <= b           # Same thing

    b.issuperset(a)  # True — b contains all of a's elements
    b >= a           # Same thing

    a.isdisjoint(b)  # False — they share elements
    # True if no common elements: a & b == set()

TODAY'S PROBLEM: Intersection of Two Arrays
================================================================================

Given two integer arrays, return an array of their intersection.
Each element in the result must be UNIQUE, and the result can be in any order.

Requirements:
    - Return only unique elements that appear in BOTH arrays
    - Use set operations for an efficient O(n + m) solution
    - Result order doesn't matter

Example:
    intersection([1, 2, 2, 1], [2, 2]) → [2]
    # 2 appears in both arrays (only return it once)

    intersection([4, 9, 5], [9, 4, 9, 8, 4]) → [4, 9] or [9, 4]
    # Both 4 and 9 appear in both arrays

    intersection([1, 2, 3], [4, 5, 6]) → []
    # No common elements

Think about:
    - What set operation gives you "elements in both"?
    - How do you convert the result back to a list?

```python
def intersection(nums1: list[int], nums2: list[int]) -> list[int]:
    """
    Find the intersection of two arrays (unique elements only).

    Args:
        nums1: First integer array
        nums2: Second integer array

    Returns:
        List of unique elements that appear in both arrays
    """
    set1 = set(nums1)
    set2 = set(nums2)
    return list(set1 & set2)


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def test(nums1, nums2, expected):
    result = intersection(nums1, nums2)
    status = "✓ PASS" if set(result) == set(expected) else "✗ FAIL"
    print(f"\nTest: intersection({nums1}, {nums2})")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: intersection()")
    print("=" * 60)

    # Test 1: Basic intersection
    test([1, 2, 2, 1], [2, 2], [2])  # Expected output is [2] (unique)

    # Test 2: Multiple common elements
    test([4, 9, 5], [9, 4, 9, 8, 4], [4, 9])  # Expected output is [4, 9] (unique)

    # Test 3: No common elements
    test([1, 2, 3], [4, 5, 6], [])  # Expected output is [] (no intersection)

    # Test 4: Identical arrays
    test([1, 2, 3], [1, 2, 3], [1, 2, 3])  # Expected output is [1, 2, 3] (all common)

    # Test 5: One array is subset
    test([1, 2], [1, 2, 3, 4], [1, 2])  # Expected output is [1, 2] (subset)

    # Test 6: Empty first array
    test([], [1, 2, 3], [])  # Expected output is [] (no elements in first array)

    # Test 7: Empty second array
    test([1, 2, 3], [], [])  # Expected output is [] (no elements in second array)

    # Test 8: Single element match
    test([1], [1], [1])  # Expected output is [1] (single common element)

    # Test 9: With negative numbers
    test([-1, 0, 1], [-1, 1, 2], [-1, 1])  # Expected output is [-1, 1] (unique common)

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n + m) - Convert both to sets, then intersect")
    print("  Space: O(n + m) - Storing both sets")
    print("=" * 60)

if __name__ == "__main__":
    run_tests()
```
