Day 2: Two-Pointer Technique - Valid Palindrome
================================================

CONCEPT: The two-pointer technique is one of the most versatile patterns 
in algorithm design. Instead of using nested loops (O(n²)), two pointers 
let you solve many problems in a single pass (O(n)).

PYTHON TIP: Multiple Assignment & Simultaneous Swaps
----------------------------------------------------
Python lets you assign multiple variables in one line and swap values 
without a temp variable:

    # Multiple assignment
    left, right = 0, len(arr) - 1

    # Simultaneous swap (no temp variable needed!)
    a, b = b, a

    # Works with list elements too
    arr[i], arr[j] = arr[j], arr[i]

    # You can even unpack in loops
    for index, value in enumerate(arr):
        print(f"Index {index} has value {value}")

COMMON TWO-POINTER PATTERNS:
1. Opposite ends → pointers start at beginning and end, move toward each other
2. Same direction → slow/fast pointers, both start at beginning
3. Two arrays → one pointer per array

APPROACH WALKTHROUGH:
    String: "A man, a plan, a canal: Panama"

    left →                              ← right
      A  m  a  n  ,     a  ...  a  m  a  P
      
    Step 1: 'A' vs 'a' (lowercased) → match! Move both inward
    Step 2: 'm' vs 'm' → match!
    Step 3: 'a' vs 'a' → match!
    ... continue until pointers cross

TIME COMPLEXITY: O(n) - single pass through string
SPACE COMPLEXITY: O(1) - only using two pointer variables

```python
def is_palindrome(s: str) -> bool:
    """
    Return True if s is a palindrome (ignoring non-alphanumeric characters and case).
    
    Use the two-pointer technique: start from both ends and work inward.
    
    Hints:
    - str.isalnum() returns True if a character is alphanumeric
    - str.lower() converts to lowercase
    - Move pointers inward, skipping non-alphanumeric characters
    """
    if not s:
        return True

    left, right = 0, len(s) - 1
    while left < right:
        # Move left pointer to the next alphanumeric character
        while left < right and not s[left].isalnum():
            left += 1
        # Move right pointer to the previous alphanumeric character
        while left < right and not s[right].isalnum():
            right -= 1
        
        # Compare characters in a case-insensitive manner
        if s[left].lower() != s[right].lower():
            return False
        
        left += 1
        right -= 1
    return True


# ============ TESTS ============
if __name__ == "__main__":
    print("Testing is_palindrome...")
    print("=" * 50)

    # Test 1: Classic palindrome with spaces and punctuation
    result = is_palindrome("A man, a plan, a canal: Panama")
    print(f"Test 1 - 'A man, a plan, a canal: Panama'")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    # Test 2: Not a palindrome
    result = is_palindrome("race a car")
    print(f"Test 2 - 'race a car'")
    print(f"  Result: {result} == False: {'✓ PASS' if result == False else '✗ FAIL'}")

    # Test 3: Empty string (edge case - considered palindrome)
    result = is_palindrome("")
    print(f"Test 3 - Empty string")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    # Test 4: Single character
    result = is_palindrome("a")
    print(f"Test 4 - 'a'")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    # Test 5: Only non-alphanumeric characters
    result = is_palindrome(",.!")
    print(f"Test 5 - ',.!'")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    # Test 6: Mixed case palindrome
    result = is_palindrome("Madam")
    print(f"Test 6 - 'Madam'")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    # Test 7: Numbers included
    result = is_palindrome("12321")
    print(f"Test 7 - '12321'")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    # Test 8: Alphanumeric mix
    result = is_palindrome("A1b2B1a")
    print(f"Test 8 - 'A1b2B1a'")
    print(f"  Result: {result} == True: {'✓ PASS' if result == True else '✗ FAIL'}")

    print("=" * 50)
    print("All tests complete!")
```
