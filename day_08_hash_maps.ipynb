CONCEPT: Hash Maps — Your O(1) Lookup Superpower
------------------------------------------------
A hash map (Python's dict) stores key-value pairs with near-instant lookups.

    # List lookup: O(n) — must scan through elements
    if target in my_list:  # Slow for large lists!

    # Dict lookup: O(1) — direct access via hashing
    if target in my_dict:  # Fast regardless of size!

This transforms many O(n²) brute-force solutions into O(n) solutions.


HOW HASH MAPS WORK (Simplified)
-------------------------------
1. A "hash function" converts your key into an index
2. The value is stored at that index
3. Looking up a key: hash it → go directly to that index

    my_dict = {"apple": 5, "banana": 3}

    "apple"  → hash() → index 42 → stores 5
    "banana" → hash() → index 17 → stores 3

    my_dict["apple"]  # Hashes "apple" → goes to index 42 → returns 5


DICT OPERATIONS AND THEIR COMPLEXITY
------------------------------------
    d = {}

    # O(1) — All the common operations are constant time!
    d[key] = value     # Insert or update
    d[key]             # Access (raises KeyError if missing)
    d.get(key)         # Access (returns None if missing)
    d.get(key, default) # Access with default value
    key in d           # Check existence
    del d[key]         # Delete

    # O(n) — These scan all elements
    list(d.keys())     # Get all keys
    list(d.values())   # Get all values
    list(d.items())    # Get all (key, value) pairs


PYTHON TIP: dict.get() vs Direct Access
---------------------------------------
Two ways to access values, with different behaviors:

    scores = {"alice": 95, "bob": 87}

    # Direct access — raises KeyError if key missing
    scores["alice"]     # → 95
    scores["charlie"]   # → KeyError: 'charlie'

    # .get() — returns None (or default) if key missing
    scores.get("alice")          # → 95
    scores.get("charlie")        # → None
    scores.get("charlie", 0)     # → 0 (custom default)

Use .get() when the key might not exist and you want to handle it gracefully.


PYTHON TIP: Building a Frequency Counter
----------------------------------------
Common pattern: count occurrences of each element.

    nums = [1, 2, 2, 3, 3, 3]

    # Method 1: Manual counting
    freq = {}
    for num in nums:
        freq[num] = freq.get(num, 0) + 1
    # freq = {1: 1, 2: 2, 3: 3}

    # Method 2: Using 'in' check
    freq = {}
    for num in nums:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1

    # Method 3: collections.Counter (most Pythonic)
    from collections import Counter
    freq = Counter(nums)  # Counter({3: 3, 2: 2, 1: 1})

All produce the same result. Counter is cleanest but understand Method 1 first!


PYTHON TIP: Iterating Over Dictionaries
---------------------------------------
    scores = {"alice": 95, "bob": 87, "charlie": 92}

    # Keys only (default)
    for name in scores:
        print(name)  # alice, bob, charlie

    # Values only
    for score in scores.values():
        print(score)  # 95, 87, 92

    # Both keys and values (most common)
    for name, score in scores.items():
        print(f"{name}: {score}")

TODAY'S PROBLEM: Two Sum
================================================================================

Given an array of integers and a target sum, return the INDICES of the two
numbers that add up to the target.

Requirements:
    - Each input has exactly one solution
    - You may not use the same element twice
    - Return the indices in any order
    - Use a hash map approach (not brute force nested loops)

Example:
    two_sum([2, 7, 11, 15], target=9) → [0, 1]
    # nums[0] + nums[1] = 2 + 7 = 9

    two_sum([3, 2, 4], target=6) → [1, 2]
    # nums[1] + nums[2] = 2 + 4 = 6

    two_sum([3, 3], target=6) → [0, 1]
    # nums[0] + nums[1] = 3 + 3 = 6

Think about:
    - For each number, what's its "complement" (the number needed to reach target)?
    - How can you use a hash map to find if the complement exists?
    - What should you store: the value or the index?


```python
def two_sum(nums: list[int], target: int) -> list[int]:
    """
    Find two numbers that add up to target and return their indices.

    Args:
        nums: List of integers
        target: Target sum

    Returns:
        List containing the two indices
    """
    if not nums:
        return []

    num_dict = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_dict:
            return [num_dict[complement], i]
        num_dict[num] = i
    return []


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================
def test(run_num: int, input_list: list[int], target: int, expected: list[int]):
    result = two_sum(input_list, target)
    status = "✓ PASS" if sorted(result) == sorted(expected) else "✗ FAIL"
    print(f"\nTest {run_num}: two_sum({input_list}, {target})")
    print(f"  Expected: {expected} (sum of {expected[0]} + {expected[1]} = {sum(expected)})")
    print(f"  Got:      {result}")
    print(f"  {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: two_sum()")
    print("=" * 60)

    # Test 1: Standard case
    test(run_num=1, input_list=[2, 7, 11, 15], target=9, expected=[0, 1])

    # Test 2: Answer in middle
    test(run_num=2, input_list=[3, 2, 4], target=6, expected=[1, 2])

    # Test 3: Same value twice
    test(run_num=3, input_list=[3, 3], target=6, expected=[0, 1])

    # Test 4: Negative numbers
    test(run_num=4, input_list=[-1, -2, -3, -4, -5], target=-8, expected=[2, 4])

    # Test 5: Mixed positive and negative
    result = two_sum([1, -2, 3, 5], 3)
    expected = [1, 4] if len(two_sum([1, -2, 3, 5], 3) or []) > 0 else [1, 3]
    result = two_sum([1, -2, 3, 5], 3)
    # Valid answers: [1, 3] (-2 + 5 = 3)
    expected = [1, 3]
    status = "✓ PASS" if sorted(result) == sorted(expected) else "✗ FAIL"
    print(f"\nTest 5: two_sum([1, -2, 3, 5], 3)")
    print(f"  Expected: {expected} (-2 + 5 = 3)")
    print(f"  Got:      {result}")
    print(f"  {status}")

    # Test 6: Larger array
    test(run_num=6, input_list=[1, 5, 8, 3, 9, 2], target=11, expected=[2, 3])

    # Test 7: Zero involved
    test(run_num=7, input_list=[0, 4, 3, 0], target=0, expected=[0, 3])

    # Test 8: First and last
    test(run_num=8, input_list=[5, 1, 1, 1, 7], target=12, expected=[0, 4])

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Single pass through array")
    print("  Space: O(n) - Hash map stores up to n elements")
    print("=" * 60)
    print("\n  Compare to brute force O(n²) with nested loops!")
    print("  Hash map lookup is the key insight.")

if __name__ == "__main__":
    run_tests()
```
