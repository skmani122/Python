CONCEPT: Big O Notation
-----------------------
Big O describes how an algorithm's runtime grows as input size (n) increases.
Think of it as answering: "If I double my input, how much longer will this take?"

Common complexities (fastest to slowest):
    O(1)        Constant    - Dictionary lookup, array access by index
    O(log n)    Logarithmic - Binary search (halving each step)
    O(n)        Linear      - Single loop through data
    O(n log n)  Log-linear  - Efficient sorting (Timsort, Mergesort)
    O(n²)       Quadratic   - Nested loops (loop inside a loop)
    O(2^n)      Exponential - Recursive Fibonacci (naive)

For interviews, focus on TIME and SPACE complexity:
    - Time:  How many operations?
    - Space: How much extra memory?


PYTHON TIP: List Operations & Their Complexity
----------------------------------------------
Knowing the cost of operations helps you write efficient code!

    nums = [1, 2, 3, 4, 5]

    # O(1) - Constant time
    nums[0]              # Access by index
    nums.append(6)       # Add to end
    nums.pop()           # Remove from end
    len(nums)            # Get length

    # O(n) - Linear time (must shift elements or scan)
    nums.insert(0, 99)   # Insert at beginning
    nums.pop(0)          # Remove from beginning
    nums.remove(3)       # Remove by value (must find it first!)
    5 in nums            # Check membership (must scan)
    nums.copy()          # Copy entire list


PYTHON TIP: enumerate() - Your New Best Friend
----------------------------------------------
When you need both INDEX and VALUE, use enumerate() instead of range(len()):

    # ❌ Clunky way
    for i in range(len(nums)):
        print(i, nums[i])

    # ✅ Pythonic way
    for i, val in enumerate(nums):
        print(i, val)

    # You can even start from a different index:
    for i, val in enumerate(nums, start=1):  # i starts at 1
        print(i, val)

TODAY'S PROBLEM: Find Maximum and Its Index
================================================================================

Write a function that finds the maximum value in a list AND returns its index.
If there are duplicates, return the index of the FIRST occurrence.

Requirements:
    - Return a tuple: (max_value, index)
    - Do NOT use the built-in max() function (practice manual traversal)
    - Handle edge case: empty list should return (None, -1)

Example:
    find_max([3, 1, 4, 1, 5, 9, 2, 6]) → (9, 5)
    find_max([7, 7, 7]) → (7, 0)  # First occurrence
    find_max([]) → (None, -1)

Think about:
    - What's the time complexity of your solution?
    - What's the space complexity?

```python
def find_max(nums: list[int]) -> tuple:
    """
    Find the maximum value and its first index in the list.

    Args:
        nums: List of integers

    Returns:
        Tuple of (max_value, index), or (None, -1) if list is empty
    """
    if not nums:
        return (None, -1)
    max_val = nums[0]
    max_idx = 0

    for idx, val in enumerate(nums):
        if val > max_val:
            max_val = val
            max_idx = idx
    return (max_val, max_idx)


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def run_tests():
    print("=" * 60)
    print("TESTING: find_max()")
    print("=" * 60)

    # Test 1: Normal case
    result = find_max([3, 1, 4, 1, 5, 9, 2, 6])
    expected = (9, 5)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest 1: find_max([3, 1, 4, 1, 5, 9, 2, 6])")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

    # Test 2: Duplicates (should return first occurrence)
    result = find_max([7, 7, 7])
    expected = (7, 0)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest 2: find_max([7, 7, 7])")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

    # Test 3: Empty list
    result = find_max([])
    expected = (None, -1)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest 3: find_max([])")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

    # Test 4: Single element
    result = find_max([42])
    expected = (42, 0)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest 4: find_max([42])")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

    # Test 5: Negative numbers
    result = find_max([-5, -2, -8, -1])
    expected = (-1, 3)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest 5: find_max([-5, -2, -8, -1])")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

    # Test 6: Max at beginning
    result = find_max([100, 1, 2, 3])
    expected = (100, 0)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest 6: find_max([100, 1, 2, 3])")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - We visit each element exactly once")
    print("  Space: O(1) - Only storing max_val and max_idx")
    print("=" * 60)

run_tests()
```
