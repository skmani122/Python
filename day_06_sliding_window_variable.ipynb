CONCEPT: Variable-Size Sliding Window
-------------------------------------
Yesterday: window size was FIXED (k elements).
Today: window size is DYNAMIC — it expands and contracts based on a condition.

The pattern:
    1. EXPAND the window by moving the right pointer
    2. Once a condition is met, SHRINK from the left to optimize
    3. Track your answer at the right moment (depends on the problem)

Variable windows solve "find the smallest/longest subarray where..." problems.


VISUAL: Smallest Subarray with Sum >= 7
---------------------------------------
Array: [2, 3, 1, 2, 4, 3]    Target: 7

    [2, 3, 1, 2, 4, 3]
     ^L ^R               sum=5  (< 7, expand)

    [2, 3, 1, 2, 4, 3]
     ^L    ^R            sum=6  (< 7, expand)

    [2, 3, 1, 2, 4, 3]
     ^L       ^R         sum=8  (≥ 7! length=4, now SHRINK)

    [2, 3, 1, 2, 4, 3]
        ^L    ^R         sum=6  (< 7, expand)

    [2, 3, 1, 2, 4, 3]
        ^L       ^R      sum=10 (≥ 7! length=4, SHRINK)

    [2, 3, 1, 2, 4, 3]
           ^L    ^R      sum=7  (≥ 7! length=3, SHRINK)

    [2, 3, 1, 2, 4, 3]
              ^L ^R      sum=6  (< 7, expand)

    [2, 3, 1, 2, 4, 3]
              ^L    ^R   sum=7  (≥ 7! length=2) ← MINIMUM!

    Answer: 2 (subarray [4, 3])


THE VARIABLE WINDOW TEMPLATE
----------------------------
    left = 0
    for right in range(len(nums)):
        # EXPAND: Add nums[right] to window

        while WINDOW_CONDITION_MET:
            # Record/update answer (if looking for minimum)
            # SHRINK: Remove nums[left] from window
            left += 1

        # Record/update answer (if looking for maximum)

Key decisions:
    - WHERE to update your answer depends on what you're optimizing
    - Finding MINIMUM length? Update INSIDE the while loop (when condition met)
    - Finding MAXIMUM length? Update OUTSIDE the while loop (before shrinking)


PYTHON TIP: while vs if for Shrinking
-------------------------------------
Use WHILE (not if) when shrinking — you might need to shrink multiple times!

    # ❌ WRONG: Only shrinks once
    if window_sum >= target:
        left += 1

    # ✅ CORRECT: Shrinks as much as possible
    while window_sum >= target:
        # update answer
        left += 1

Why while? The window might be much larger than needed. Example:
    Array: [1, 1, 1, 1, 100]  Target: 7
    After adding 100, sum = 104. Need to shrink 4 times, not just once!


PYTHON TIP: min() with Default for Empty Cases
-----------------------------------------------
When tracking minimum length, initialize with infinity:

    min_length = float('inf')

    # At the end, check if we ever found a valid window
    return min_length if min_length != float('inf') else 0

Alternative using min() with a default:

    # If candidates is empty, return 0
    return min(candidates, default=0)


PYTHON TIP: Updating Min/Max Inline
-----------------------------------
Clean one-liners for tracking best values:

    # Track minimum
    min_len = min(min_len, right - left + 1)

    # Track maximum
    max_len = max(max_len, right - left + 1)

Remember: Window length = right - left + 1 (inclusive of both endpoints)

TODAY'S PROBLEM: Minimum Size Subarray Sum
================================================================================

Given an array of positive integers and a target sum, find the MINIMAL LENGTH
of a contiguous subarray whose sum is greater than or equal to the target.
If no such subarray exists, return 0.

Requirements:
    - Find the shortest subarray with sum >= target
    - All numbers are positive (this is important for the algorithm!)
    - Return 0 if no valid subarray exists

Example:
    min_subarray_len(7, [2, 3, 1, 2, 4, 3]) → 2
    # Subarray [4, 3] has sum=7 and length=2

    min_subarray_len(4, [1, 4, 4]) → 1
    # Subarray [4] has sum=4 and length=1

    min_subarray_len(11, [1, 1, 1, 1, 1]) → 0
    # No subarray sums to 11

Think about:
    - When do you expand the window?
    - When do you shrink the window?
    - Where do you update the minimum length?

```python
def min_subarray_len(target: int, nums: list[int]) -> int:
    """
    Find the minimal length of a subarray with sum >= target.

    Args:
        target: The target sum
        nums: List of positive integers

    Returns:
        Length of smallest subarray with sum >= target, or 0 if none exists
    """

    if not nums:
        return 0

    left = 0
    current_sum = 0
    min_length = float('inf')

    for right in range(len(nums)):
        current_sum += nums[right]

        while current_sum >= target:
            min_length = min(min_length, right - left + 1)
            current_sum -= nums[left]
            left += 1
            if left > right:  # Just a safety check, should not happen with positive integers
                break

    return min_length if min_length != float('inf') else 0


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def test(run_num: int, input_list: list[int], k: int, expected: int):
    result = min_subarray_len(k, input_list)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"\nTest {run_num}: min_subarray_len({k}, {input_list})")
    print(f"  Expected: {expected}")
    print(f"  Got:      {result}")
    print(f"  {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: min_subarray_len()")
    print("=" * 60)

    # Test 1: Standard case
    test(run_num=1, input_list=[2, 3, 1, 2, 4, 3], k=7, expected=2)

    # Test 2: Single element satisfies
    test(run_num=2, input_list=[1, 4, 4], k=4, expected=1)

    # Test 3: No valid subarray
    test(run_num=3, input_list=[1, 1, 1, 1, 1], k=11, expected=0)

    # Test 4: Entire array needed
    test(run_num=4, input_list=[1, 2, 3, 4, 5], k=15, expected=5)

    # Test 5: Multiple valid, pick shortest
    test(run_num=5, input_list=[2, 2, 2, 6, 2], k=6, expected=1)

    # Test 6: Target is 0 (any element works)
    test(run_num=6, input_list=[1, 2, 3], k=0, expected=1)

    # Test 7: Empty array
    test(run_num=7, input_list=[], k=5, expected=0)

    # Test 8: Large element at end
    test(run_num=8, input_list=[1, 2, 3, 4, 100], k=100, expected=1)

    # Test 9: All same values
    test(run_num=9, input_list=[3, 3, 3, 3, 3], k=9, expected=3)

    # Test 10: Exact match requires shrinking multiple times
    test(run_num=10, input_list=[1, 1, 1, 1, 7], k=7, expected=1)

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Each element added and removed at most once")
    print("  Space: O(1) - Only tracking sum, pointers, and min_length")
    print("=" * 60)
    print("\n  Note: Even though there's a while loop inside the for loop,")
    print("  each element is visited at most twice (once by right, once by left).")
    print("  This is a key insight for variable sliding window complexity!")

run_tests()
```
