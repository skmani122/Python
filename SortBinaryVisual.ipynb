```python
import random
import numpy as np
from IPython.display import HTML

# Run this cell in a Jupyter Notebook to visualize Binary Insertion Sort.

import matplotlib.pyplot as plt
import matplotlib.animation as animation

def binary_search(arr, val, start, end):
    while start <= end:
        mid = (start + end) // 2
        if arr[mid] == val:
            return mid + 1
        elif arr[mid] < val:
            start = mid + 1
        else:
            end = mid - 1
    return start

def binary_insertion_sort_states(a):
    arr = a.copy()
    states = []
    states.append((arr.copy(), None, None, None))  # initial
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        pos = binary_search(arr, key, 0, i - 1)
        j = i
        # shift right one by one, recording each step
        while j > pos:
            arr[j] = arr[j - 1]
            states.append((arr.copy(), j, pos, i))  # (array, moved_index, target_pos, current_i)
            j -= 1
        arr[pos] = key
        states.append((arr.copy(), pos, pos, i))  # insertion final step
    states.append((arr.copy(), None, None, None))  # final
    return states

def animate_sort(states, interval=200, figsize=(10,4)):
    fig, ax = plt.subplots(figsize=figsize)
    plt.close(fig)  # prevent double display in notebooks; we'll show HTML below

    arr0 = states[0][0]
    x = np.arange(len(arr0))
    bars = ax.bar(x, arr0, align='center', color='lightblue')
    ax.set_title('Binary Insertion Sort')
    ax.set_xlim(-0.5, len(arr0)-0.5)
    ax.set_ylim(0, max(arr0)*1.1)

    def update(frame):
        arr, moved_idx, pos, cur_i = states[frame]
        for rect, h in zip(bars, arr):
            rect.set_height(h)
            rect.set_color('lightblue')
        if moved_idx is not None:
            bars[moved_idx].set_color('red')   # element being shifted
        if pos is not None:
            bars[pos].set_color('green')       # insertion position
        if cur_i is not None:
            bars[cur_i].set_color('orange')    # current key index
        ax.set_title(f'Frame {frame+1}/{len(states)}')
        return bars

    anim = animation.FuncAnimation(fig, update, frames=len(states), interval=interval, blit=False)
    return HTML(anim.to_jshtml())

# Example usage: create random array and show animation
if __name__ == "__main__":
    sample = random.sample(range(1, 100), 20)
    states = binary_insertion_sort_states(sample)
    # In a Jupyter Notebook cell, call animate_sort(states)
    # For direct script run, save animation as GIF (requires imagemagick/ffmpeg)
    html_anim = animate_sort(states, interval=150)
    display(html_anim)
```
