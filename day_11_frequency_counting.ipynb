CONCEPT: Frequency Counting — A Fundamental Pattern
---------------------------------------------------
Many problems require counting how often each element appears.
Python's collections module makes this elegant and efficient.

The pattern: "How many times does each element occur?"

    [1, 2, 2, 3, 3, 3]  →  {1: 1, 2: 2, 3: 3}
    "hello"            →  {'h': 1, 'e': 1, 'l': 2, 'o': 1}


COLLECTIONS.COUNTER — The Counting Specialist
---------------------------------------------
Counter is a dict subclass designed specifically for counting:

    from collections import Counter

    # Count from iterable
    nums = [1, 2, 2, 3, 3, 3]
    freq = Counter(nums)  # Counter({3: 3, 2: 2, 1: 1})

    # Count characters in string
    word = "mississippi"
    freq = Counter(word)  # Counter({'i': 4, 's': 4, 'p': 2, 'm': 1})

    # Access counts (returns 0 for missing keys, not KeyError!)
    freq['i']  # 4
    freq['z']  # 0 (not KeyError!)


COUNTER'S SPECIAL METHODS
-------------------------
    from collections import Counter

    c = Counter("abracadabra")
    # Counter({'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1})

    # Most common elements (sorted by count)
    c.most_common(2)    # [('a', 5), ('b', 2)]
    c.most_common()     # All elements, sorted by frequency

    # Total count
    c.total()           # 11 (Python 3.10+)
    sum(c.values())     # 11 (works in all versions)

    # Arithmetic with Counters!
    c1 = Counter("aab")  # {'a': 2, 'b': 1}
    c2 = Counter("abb")  # {'a': 1, 'b': 2}
    c1 + c2              # Counter({'a': 3, 'b': 3})
    c1 - c2              # Counter({'a': 1})  (keeps only positive)


COLLECTIONS.DEFAULTDICT — The Auto-Initializer
----------------------------------------------
defaultdict automatically creates missing keys with a default value:

    from collections import defaultdict

    # Regular dict — KeyError if key missing
    d = {}
    d["missing"]  # KeyError!

    # defaultdict(int) — missing keys default to 0
    d = defaultdict(int)
    d["missing"]    # 0 (auto-created!)
    d["count"] += 1 # Works! No need to check if key exists

    # defaultdict(list) — missing keys default to []
    d = defaultdict(list)
    d["fruits"].append("apple")  # Auto-creates the list!

    # defaultdict(set) — missing keys default to set()
    d = defaultdict(set)
    d["tags"].add("python")


PYTHON TIP: Manual Counting vs Counter
--------------------------------------
    text = "hello world"

    # Manual counting (understand this first!)
    freq = {}
    for char in text:
        freq[char] = freq.get(char, 0) + 1

    # With defaultdict (cleaner)
    from collections import defaultdict
    freq = defaultdict(int)
    for char in text:
        freq[char] += 1

    # With Counter (cleanest!)
    from collections import Counter
    freq = Counter(text)

All produce: {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}


PYTHON TIP: Counter Comparisons
-------------------------------
Two Counters are equal if they have the same counts:

    Counter("abc") == Counter("cba")   # True! Same counts
    Counter("abc") == Counter("abcc")  # False! Different counts

This makes anagram checking trivial:

    def is_anagram(s1, s2):
        return Counter(s1) == Counter(s2)

TODAY'S PROBLEM: First Unique Character in a String
================================================================================

Given a string s, find the FIRST non-repeating character and return its index.
If no unique character exists, return -1.

Requirements:
    - Use frequency counting (Counter preferred)
    - Return the INDEX of the first unique character
    - Lowercase English letters only

Example:
    first_uniq_char("leetcode") → 0
    # 'l' appears once, and it's at index 0

    first_uniq_char("loveleetcode") → 2
    # 'l' appears twice, 'o' appears twice, 'v' appears once at index 2

    first_uniq_char("aabb") → -1
    # No unique characters

Think about:
    - How do you count all character frequencies?
    - After counting, how do you find the FIRST unique one?
    - Why do you need TWO passes through the string?


```python
from collections import Counter, defaultdict

def first_uniq_char(s: str) -> int:
    """
    Find the index of the first non-repeating character.

    Args:
        s: String of lowercase letters

    Returns:
        Index of first unique character, or -1 if none exists
    """
    freq = Counter(s)

    for i, char in enumerate(s):
        if freq[char] == 1:
            return i

    return -1


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================
def test(input, expected):
    result = first_uniq_char(input)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"Input: '{input}' | Expected: {expected} | Got: {result} | {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: first_uniq_char()")
    print("=" * 60)

    # Test 1: First character is unique
    test("leetcode", 0)  # 'l' is unique at index 0

    # Test 2: Unique character in middle
    test("loveleetcode", 2)  # 'v' is unique at index 2

    # Test 3: No unique characters
    test("aabb", -1)  # All characters repeat

    # Test 4: Single character
    test("a", 0)  # 'a' is unique at index 0

    # Test 5: Empty string
    test("", -1)  # No characters, so no unique character

    # Test 6: All same character
    test("aaaa", -1)  # All 'a's repeat, no unique character

    # Test 7: Unique at end
    test("aabbccd", 6)  # 'd' is unique at index 6

    # Test 8: Multiple unique, return first
    test("abcabc", -1)  # All characters repeat, no unique character

    # Test 9: First of multiple uniques
    test("abcdefabc", 3)  # 'd' is first unique at index 3

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Two passes through string (count + find)")
    print("  Space: O(1) - At most 26 characters in frequency map")
    print("=" * 60)
    print("\n  Note: Space is O(1) because alphabet size is constant.")
    print("  For Unicode, it would be O(k) where k = unique chars.")

if __name__ == "__main__":
    run_tests()
```

BONUS PROBLEM: Ransom Note
==============================================================================

Given two strings ransomNote and magazine, return True if ransomNote can be
constructed from the letters in magazine. Each letter can only be used once.

### Example:
- can_construct("aa", "aab") → True  (magazine has 2 a's)
- can_construct("aa", "ab") → False  (magazine only has 1 a)

- Hint: Count letters in magazine, then "spend" them for ransomNote

```python

def can_construct(ransomNote: str, magazine: str) -> bool:
    '''Check if ransomNote can be made from magazine letters.'''
    # YOUR CODE HERE
    magazine_count = Counter(magazine)
    for char in ransomNote:
        if magazine_count[char] <= 0:
            return False
        magazine_count[char] -= 1
    return True

print("Fail" if can_construct("a", "b") else "Pass")  # False
print("Fail" if can_construct("aa", "ab") else "Pass")  # False
print("Pass" if can_construct("aa", "aab") else "Fail")  # True
```
