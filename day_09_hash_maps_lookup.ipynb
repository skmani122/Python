CONCEPT: The Power of O(1) Existence Checks
-------------------------------------------
Yesterday we used hash maps to find complements in Two Sum.
Today we focus on the core superpower: checking "does X exist?" in O(1) time.

Compare these approaches for checking if an element exists:

    # List: O(n) — must scan through
    if target in my_list:
        ...

    # Dict: O(1) — direct lookup
    if target in my_dict:
        ...

This difference is HUGE for large datasets or when checking many times.


WHEN TO "PRE-PROCESS" INTO A HASH MAP
-------------------------------------
Pattern: If you need to check existence/lookup MULTIPLE times, it's often
worth building a hash map first, even if building takes O(n).

    # Bad: O(n) lookup × m queries = O(n × m)
    for query in queries:
        if query in large_list:  # O(n) each time!
            ...

    # Good: O(n) build + O(1) lookup × m queries = O(n + m)
    lookup_set = set(large_list)  # Build once: O(n)
    for query in queries:
        if query in lookup_set:   # O(1) each time!
            ...


PYTHON TIP: Dict as a "Seen" Tracker
------------------------------------
Common pattern: track elements you've encountered with their positions.

    seen = {}  # {value: index}

    for i, num in enumerate(nums):
        if num in seen:
            # Found duplicate! seen[num] gives the first occurrence index
            print(f"Duplicate {num} at indices {seen[num]} and {i}")
        else:
            seen[num] = i  # First time seeing this value

This "seen" dict pattern appears in MANY problems:
    - Two Sum (store value → index)
    - Contains Duplicate (just check if seen before)
    - First repeated character
    - Find duplicate within distance k


PYTHON TIP: Dict Comprehensions
-------------------------------
Create dicts in one line, similar to list comprehensions:

    # List comprehension
    squares_list = [x**2 for x in range(5)]  # [0, 1, 4, 9, 16]

    # Dict comprehension
    squares_dict = {x: x**2 for x in range(5)}  # {0:0, 1:1, 2:4, 3:9, 4:16}

    # With condition
    even_squares = {x: x**2 for x in range(10) if x % 2 == 0}

    # Swap keys and values
    original = {"a": 1, "b": 2}
    flipped = {v: k for k, v in original.items()}  # {1: "a", 2: "b"}

    # From two lists (use zip!)
    keys = ["name", "age", "city"]
    vals = ["Alice", 30, "NYC"]
    person = dict(zip(keys, vals))  # {"name": "Alice", "age": 30, "city": "NYC"}


PYTHON TIP: The "in" Operator with Dicts
----------------------------------------
Be clear about what you're checking:

    d = {"apple": 5, "banana": 3}

    # Checks KEYS (most common, default behavior)
    "apple" in d        # True
    "cherry" in d       # False

    # Checks VALUES (need .values())
    5 in d.values()     # True
    10 in d.values()    # False

    # Check both key AND value
    ("apple", 5) in d.items()  # True

TODAY'S PROBLEM: Contains Duplicate II
================================================================================

Given an integer array nums and an integer k, return True if there are two
DISTINCT indices i and j such that:
    - nums[i] == nums[j] (same value)
    - abs(i - j) <= k (indices are within k distance)

In other words: is there a duplicate within a "window" of size k?

Requirements:
    - Use a hash map to achieve O(n) time complexity
    - Return True if such a pair exists, False otherwise

Example:
    contains_nearby_duplicate([1, 2, 3, 1], k=3) → True
    # nums[0] = nums[3] = 1, and |0 - 3| = 3 <= 3 ✓

    contains_nearby_duplicate([1, 0, 1, 1], k=1) → True
    # nums[2] = nums[3] = 1, and |2 - 3| = 1 <= 1 ✓

    contains_nearby_duplicate([1, 2, 3, 1, 2, 3], k=2) → False
    # Duplicates exist, but none are within distance 2

Think about:
    - What should you store in the hash map? Value → ???
    - When you see a value again, what do you check?
    - Should you update the stored index? When?


```python
def contains_nearby_duplicate(nums: list[int], k: int) -> bool:
    """
    Check if there are duplicates within k distance of each other.

    Args:
        nums: List of integers
        k: Maximum allowed distance between duplicates

    Returns:
        True if duplicate pair exists within distance k, False otherwise
    """
    seen = {}  # Dictionary to store the last index of each number

    for i, num in enumerate(nums):
        if num in seen:
            if i - seen[num] <= k:  # Check if the distance is within k
                return True
        seen[num] = i  # Update the last index of the number

    return False  # No duplicates found within distance k


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def test(input, k, expected):
    result = contains_nearby_duplicate(input, k)
    status = "✓ PASS" if result == expected else "✗ FAIL"
    print(f"Input: {input}  Expected: {expected}  Got: {result}\n  {status}")

def run_tests():
    print("=" * 60)
    print("TESTING: contains_nearby_duplicate()")
    print("=" * 60)

    # Test 1: Duplicate within range
    test([1, 2, 3, 1], 3, True)  # indices 0 and 3 differ by 3

    # Test 2: Adjacent duplicates
    test([1, 0, 1, 1], 1, True)  # indices 2 and 3 differ by 1

    # Test 3: Duplicates exist but too far apart
    test([1, 2, 3, 1, 2, 3], 2, False)  # duplicates are 3 apart, k=2

    # Test 4: No duplicates at all
    test([1, 2, 3, 4, 5], 3, False)

    # Test 5: k=0 (must be same index, impossible for distinct)
    test([1, 2, 1], 0, False)

    # Test 6: Empty array
    test([], 1, False)

    # Test 7: Single element
    test([1], 1, False)

    # Test 8: Large k (bigger than array)
    test([1, 2, 1], 10, True)  # indices 0 and 2 differ by 2, k=10 allows it

    # Test 9: Multiple duplicates, check closest
    test([1, 2, 3, 4, 1, 1], 2, True)  # indices 4 and 5 differ by 1, k=2 allows it

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Single pass through array")
    print("  Space: O(min(n, k)) - Hash map stores at most n elements")
    print("=" * 60)

if __name__ == "__main__":
    run_tests()
```
