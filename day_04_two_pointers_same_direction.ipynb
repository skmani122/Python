CONCEPT: Fast/Slow Pointers (Same Direction)
--------------------------------------------
Yesterday we used pointers moving TOWARD each other from opposite ends.
Today: both pointers start at the SAME end and move in the SAME direction,
but at DIFFERENT speeds or under DIFFERENT conditions.

Common patterns:

    1. SLOW/FAST for partitioning:
       - Slow pointer: marks the "boundary" of processed/valid elements
       - Fast pointer: scans ahead looking for elements to process

       [processed | unprocessed...]
              ^slow        ^fast

    2. TORTOISE & HARE (different speeds):
       - Slow moves 1 step, fast moves 2 steps
       - Used for cycle detection (we'll see this with linked lists in Week 4)

When to use same-direction two pointers:
    ✓ "In-place" modifications (no extra array allowed)
    ✓ Partitioning arrays (move elements that match condition)
    ✓ Removing duplicates from sorted arrays
    ✓ Cycle detection (linked lists, repeated sequences)


VISUAL: Move Zeroes Example
---------------------------
Goal: Move all 0s to the end while maintaining order of non-zeros

    [0, 1, 0, 3, 12]
     ^slow
     ^fast

    Fast finds 1 (non-zero), swap with slow, advance both:
    [1, 0, 0, 3, 12]
        ^slow
           ^fast

    Fast finds 3, swap with slow:
    [1, 3, 0, 0, 12]
           ^slow
               ^fast

    Fast finds 12, swap with slow:
    [1, 3, 12, 0, 0]  ✓ Done!


PYTHON TIP: Elegant Swapping
----------------------------
Python lets you swap values without a temp variable using tuple unpacking:

    # ❌ Other languages need a temp variable
    temp = a
    a = b
    b = temp

    # ✅ Python's elegant swap
    a, b = b, a

    # Works with list elements too!
    nums[i], nums[j] = nums[j], nums[i]

    # Even works with multiple values
    a, b, c = c, a, b  # Rotate values!

How it works: Python evaluates the RIGHT side first (creating a tuple),
then unpacks it into the LEFT side. No temporary variable needed.


PYTHON TIP: Modifying Lists In-Place
------------------------------------
Many interview problems require "in-place" modification (O(1) extra space).
This means you modify the original list rather than creating a new one.

    # ❌ Creates new list (O(n) space)
    nums = [x for x in nums if x != 0]

    # ✅ Modifies in place (O(1) space)
    # Use index manipulation, swapping, etc.

Why in-place matters:
    - Memory efficiency for large datasets
    - Sometimes required by the problem constraints
    - Shows interviewer you understand space complexity

================================================================================
TODAY'S PROBLEM: Move Zeroes
================================================================================

Given an integer array nums, move all 0's to the end while maintaining
the relative order of the non-zero elements.

Requirements:
    - Modify the array IN-PLACE (don't create a new array)
    - Maintain the relative order of non-zero elements
    - Return None (modify nums directly)

Example:
    nums = [0, 1, 0, 3, 12]
    move_zeroes(nums)
    # nums is now [1, 3, 12, 0, 0]

    nums = [0, 0, 1]
    move_zeroes(nums)
    # nums is now [1, 0, 0]

Think about:
    - What does the slow pointer represent?
    - When should you swap vs just advance?
    - What's the time and space complexity?

```python
def move_zeroes(nums: list[int]) -> None:
    """
    Move all zeros to the end of the array in-place.

    Args:
        nums: List of integers (modified in-place)

    Returns:
        None (modifies nums directly)
    """
    slow = fast = 0
    while fast < len(nums):
        if nums[fast] != 0:
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1
        fast += 1


# ==============================================================================
# TESTS - Run this file to check your solution
# ==============================================================================

def run_tests():
    print("=" * 60)
    print("TESTING: move_zeroes()")
    print("=" * 60)

    # Test 1: Mixed zeros
    nums = [0, 1, 0, 3, 12]
    move_zeroes(nums)
    expected = [1, 3, 12, 0, 0]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 1: [0, 1, 0, 3, 12]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 2: Zeros at start
    nums = [0, 0, 1]
    move_zeroes(nums)
    expected = [1, 0, 0]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 2: [0, 0, 1]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 3: No zeros
    nums = [1, 2, 3]
    move_zeroes(nums)
    expected = [1, 2, 3]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 3: [1, 2, 3]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 4: All zeros
    nums = [0, 0, 0]
    move_zeroes(nums)
    expected = [0, 0, 0]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 4: [0, 0, 0]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 5: Single element
    nums = [0]
    move_zeroes(nums)
    expected = [0]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 5: [0]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 6: Empty list
    nums = []
    move_zeroes(nums)
    expected = []
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 6: []")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 7: Zero at end (already correct)
    nums = [1, 2, 0]
    move_zeroes(nums)
    expected = [1, 2, 0]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 7: [1, 2, 0]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    # Test 8: Alternating
    nums = [0, 1, 0, 2, 0, 3]
    move_zeroes(nums)
    expected = [1, 2, 3, 0, 0, 0]
    status = "✓ PASS" if nums == expected else "✗ FAIL"
    print(f"\nTest 8: [0, 1, 0, 2, 0, 3]")
    print(f"  Expected: {expected}")
    print(f"  Got:      {nums}")
    print(f"  {status}")

    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)
    print("  Time:  O(n) - Single pass through the array")
    print("  Space: O(1) - Only two pointers, no extra array")
    print("=" * 60)

run_tests()
```
